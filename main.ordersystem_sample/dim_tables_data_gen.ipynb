```python
%pip install faker
# %restart_python

import uuid
from faker import Faker
from datetime import datetime
import random
import json

fake = Faker()

# -------------------------------------------
# Customers (25 inserts)
# -------------------------------------------
for _ in range(25):
    address = json.dumps({
        "street": fake.street_address(),
        "city": fake.city(),
        "state": fake.state_abbr(),
        "zip": fake.zipcode(),
        "country": "USA"
    })
    customer_id = str(uuid.uuid4())
    name = fake.name().replace("'", "''")  # Escape single quotes for SQL
    email = fake.email()
    phone = fake.phone_number().replace("'", "''")
    is_deleted = random.choice([False, False, False, True])
    now = datetime.now().isoformat()

    spark.sql(f"""
        INSERT INTO main.ordersystem_sample.customers
        VALUES (
            '{customer_id}',
            '{name}',
            '{email}',
            '{phone}',
            '{address}',
            {str(is_deleted).lower()},
            TIMESTAMP('{now}'),
            TIMESTAMP('{now}')
        )
    """)

# -------------------------------------------
# Stores (5 inserts)
# -------------------------------------------
for i in range(5):
    store_id = f"store_{i+1}"
    store_name = fake.company().replace("'", "''")
    location = fake.city().replace("'", "''")

    spark.sql(f"""
        INSERT INTO main.ordersystem_sample.stores
        VALUES (
            '{store_id}',
            '{store_name}',
            '{location}'
        )
    """)

# -------------------------------------------
# Products (10 inserts)
# -------------------------------------------
for i in range(10):
    product_id = f"product_{i+1}"
    product_name = fake.word().replace("'", "''")
    price = round(random.uniform(5, 500), 2)

    spark.sql(f"""
        INSERT INTO main.ordersystem_sample.products
        VALUES (
            '{product_id}',
            '{product_name}',
            {price}
        )
    """)

print("✅ Rows inserted directly via single-row SQL INSERT statements into Delta tables.")
```
