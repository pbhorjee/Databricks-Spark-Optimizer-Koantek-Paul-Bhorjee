```python
%python


most_sold_df = spark.sql("""
SELECT 
  p.product_id,
  p.product_name,
  SUM(oli.quantity) AS total_quantity_sold
FROM main.ordersystem_sample.order_line_items oli
JOIN main.ordersystem_sample.orders o
  ON oli.order_id = o.order_id
JOIN main.ordersystem_sample.products p
  ON oli.product_id = p.product_id
WHERE year(o.created_at) = year(current_date())
  AND month(o.created_at) = month(current_date())
  AND oli.item_status IN ('DELIVERED', 'SHIPPED')
GROUP BY p.product_id, p.product_name
ORDER BY total_quantity_sold DESC
LIMIT 1
""")

most_sold_df.show()
```

```python
most_sold_lastmonth_df = spark.sql("""
SELECT 
  p.product_id,
  p.product_name,
  SUM(oli.quantity) AS total_quantity_sold
FROM main.ordersystem_sample.order_line_items oli
JOIN main.ordersystem_sample.orders o
  ON oli.order_id = o.order_id
JOIN main.ordersystem_sample.products p
  ON oli.product_id = p.product_id
WHERE o.created_at >= date_trunc('month', current_date - INTERVAL 1 month)
  AND o.created_at < date_trunc('month', current_date)
  AND oli.item_status IN ('DELIVERED', 'SHIPPED')
GROUP BY p.product_id, p.product_name
ORDER BY total_quantity_sold DESC
LIMIT 1;
""")

most_sold_lastmonth_df.show()
```
